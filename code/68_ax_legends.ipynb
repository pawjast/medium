# Handling `ax.legend()` in matplotlib

## Imports

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Mock data

```python
nr_of_points = 10
np.random.seed(2)
X = list(range(nr_of_points))
Y_1 = np.random.randn(nr_of_points)
Y_2 = np.random.rand(nr_of_points)
Y_3 = Y_1 + Y_2
```

## Plot default legend

### Basic approach

If you add `label` parameter inside plotting function, then you can generate `legend` without any parameters.

```python
fig, ax = plt.subplots(
    figsize=(16, 8),
)
ax.set_facecolor("ivory")

ax.plot(
    X, Y_1,
    marker="d",
    color="red",
    label="Data 1"
)

ax.plot(
    X, Y_2,
    marker="o",
    color="green",
    label="Data 2"
)

ax.plot(
    X, Y_3,
    marker="s",
    color="blue",
    label="Data 1 + Data 2"
)

ax.legend();
```

Otherwise you'll get a warning there's nothing to add to the legend.

```python
fig, ax = plt.subplots(
    figsize=(16, 8),
)
ax.set_facecolor("ivory")

ax.plot(
    X, Y_1,
    marker="d",
    color="red",
)

ax.plot(
    X, Y_2,
    marker="o",
    color="green",
)

ax.plot(
    X, Y_3,
    marker="s",
    color="blue",
)

ax.legend();
```

### Alternative approach

Explicitly specify which lines get shown in the legend.

```python
fig, ax = plt.subplots(
    figsize=(16, 8),
)
ax.set_facecolor("ivory")

data1, = ax.plot(
    X, Y_1,
    marker="d",
    color="red",
)

data2, = ax.plot(
    X, Y_2,
    marker="o",
    color="green",
)

data_all, = ax.plot(
    X, Y_3,
    marker="s",
    color="blue",
)

ax.legend(
    handles=[data1, data2, data_all],
    labels=["Data 1", "Data 2", "Data 1 + Data 2"],
);
```

## Move the legend outside of axis

We need to use these properties:
* `loc`
* `bbox_to_anchor`

### Missing `loc`

```python
fig, ax = plt.subplots(
    figsize=(16, 8),
)
ax.set_facecolor("ivory")

ax.plot(
    X, Y_1,
    marker="d",
    color="red",
    label="Data 1"
)

ax.plot(
    X, Y_2,
    marker="o",
    color="green",
    label="Data 2"
)

ax.plot(
    X, Y_3,
    marker="s",
    color="blue",
    label="Data 1 + Data 2"
)

# Without setting `loc`, it's chosen randomly so the result is unpredictable

ax.legend(
    title="Legend",
    bbox_to_anchor=(1, 0.1)
);
```

```python
ax.get_legend()._loc

# 0 -> best location, so it can be random
```

```python
fig, ax = plt.subplots(
    figsize=(16, 8),
)
ax.set_facecolor("ivory")

ax.plot(
    X, Y_1,
    marker="d",
    color="red",
    label="Data 1"
)

ax.plot(
    X, Y_2,
    marker="o",
    color="green",
    label="Data 2"
)

ax.plot(
    X, Y_3,
    marker="s",
    color="blue",
    label="Data 1 + Data 2"
)

# Without setting `loc`, it's chosen randomly so the result is unpredictable

ax.legend(
    title="Legend",
    loc="lower center",
    bbox_to_anchor=(0.5, 1.01),
    borderaxespad=0,
    ncols=3,
    fontsize=12,
    labelcolor=["red", "green", "blue"],
    title_fontproperties=dict(
        size=20
    )
);

# many more can be tweaked
```
