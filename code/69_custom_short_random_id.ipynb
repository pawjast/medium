## Actual solution

```python
import string
import random
```

```python
# Get letters and numbers
string.ascii_letters + string.digits
```

```python
# Generate 4 random chars with replacement
random.choices(
    population=string.ascii_letters + string.digits,
    k=4
)
```

```python
# Combine them
appendix = ''.join(
    random.choices(string.ascii_letters + string.digits, k=4)
)

appendix
```

```python
def count_duplicates(length: int, total_runs: int = 1_000_000) -> dict:
    characters = string.ascii_letters + string.digits
    
    # Generate `total_runs` number of strings of `length`
    generated_strings = [
        ''.join(random.choices(characters, k=length)) 
        for _ in range(total_runs)
    ]
    unique_count = len(set(generated_strings))
    duplicate_count = total_runs - unique_count
    
    return {
        "length": length,
        "total_runs": total_runs,
        "unique": unique_count,
        "duplicates": duplicate_count
    }
```

```python
for id_length in [3, 4, 5]:
    print(f"\n--- String Length: {id_length} ---")
    for runs in [100, 200, 400, 800, 1600, 3200]:
        results = count_duplicates(length=id_length, total_runs=runs)
        print(
            f"No of runs {runs:>5,}: {results['unique']:>6,} unique, "
            f"{results['duplicates']:,} duplicates"
        )
```

## Check duplicates in 10,000 trials drawing 100 samples each trial with 4 char strings

```python
length = 3
sample_sizes = [25, 50, 75, 100, 200]
repetition_counts = [100, 1_000, 10_000]

for size in sample_sizes:
    print(f"\n--- Sample Size: {size} ---")
    
    for num_trials in repetition_counts:
        runs_with_duplicates = 0
        
        for _ in range(num_trials):
            results = count_duplicates(length=length, total_runs=size)
            if results["duplicates"] > 0:
                runs_with_duplicates += 1
                
        percentage = (runs_with_duplicates / num_trials) * 100
        print(
            f"{runs_with_duplicates:>3,} of {num_trials:>6,} runs "
            f"had at least 1 duplicate ({percentage:>4.2f}%)"
        )
```
