# Zipping and Long Zipping in Python

## Imports

```python
from itertools import zip_longest
```

## Example 1: just zipping

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel"]
```

### Create a zip object

```python
zip_obj = zip(features, animals)
```

```python
zip_obj
```

```python
type(zip_obj)
```

`zip` object is an iterator.

This means:

* it knows it's current position
* it gives one item at a time
* it's exhaustive - once you used it, it's gone

### Get items from zip object

```python
next(zip_obj)
```

```python
next(zip_obj)
```

```python
next(zip_obj)
```

```python
next(zip_obj)
```

```python
next(zip_obj)
```

### A better way to get items from zip object

```python
# First we need to recreate the zip object since we've already exhausted it
zip_obj = zip(features, animals)
```

```python
for item in zip_obj:
    print(item)
```

```python
# Or get individual items from respective lists
for feat, animal in zip(features, animals):
    print(f"{feat} {animal}")
```

### Best way to get the items from zip object

```python
# recreate the zip object
zip_obj = zip(features, animals)
```

```python
zip_items = list(zip_obj)
```

```python
zip_items
```

`list` is iterable. Iterable stores all values in memory. It can be looped over and over again. It can be created once and used forever.


## Example 2: dealng with uneven length lists

This works and the output is truncated to the shortest list.

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
```

```python
for item in zip(features, animals):
    print(item)
```

## Example 3: dealng with uneven length lists with warning

Using `strict` parameter.

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
```

```python
for item in zip(features, animals, strict=True):
    print(item)
```

## Example 4: dealing with unevent list using default value

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
```

```python
for item in zip_longest(features, animals):
    print(item)
```

```python
for item in zip_longest(features, animals, fillvalue="???"):
    print(item)
```

## Example 5: working with mulitple inputs

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
odd_number = [13, 27, 81]
```

```python
for item in zip(features, animals, odd_number):
    print(item)
```

```python
for item in zip_longest(features, animals, odd_number, fillvalue="??"):
    print(item)
```

## Converting zip to other containers

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy", "Macho"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
odd_number = [53, 13, 27, 81, 3]
```

```python
my_list = list(zip(features, animals, odd_number))
my_list
```

```python
my_tuple = tuple(zip(features, animals, odd_number))
my_tuple
```

```python
my_dict = dict(zip(animals, features))
my_dict
```

### Comprehension example

```python
my_dict = {
    animal: [feature, number]
    for feature, animal, number in zip(features, animals, odd_number)
}

my_dict
```

## (un)Zipping with unpacking operator

```python
features = ["Sneezing", "Dancing", "Grumpy", "Sleepy"]
animals = ["Llama", "Beaver", "Monkey", "Squirrel", "Bobcat"]
```

```python
zipped = list(zip(features, animals))
zipped
```

```python
f, a = zip(*zipped)
```
